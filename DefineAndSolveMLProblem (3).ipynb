```python
# YOUR CODE HERE
print(f"Info{df.info}")

print("\nSummary statistics for columns:")
display(df.describe())
display(df.dtypes)

print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
numberCols = df.select_dtypes(include=['int64', 'float64'])
catCols = df.select_dtypes(include=['object', 'bool'])
# print(numberCols.columns.tolist())
# print(catCols.columns.tolist())

#removing missing values from numerical data points
for col in numberCols:
    medianOfCol = df[col].median()
    df[col].fillna(medianOfCol, inplace=True)

#removing missing values from categorical data points
for col in catCols:
    modeOfCol = df[col].mode()[0]
    df[col].fillna(modeOfCol, inplace=True)

#Now we check for new count
print(df.isnull().sum())
```

```python
# Features to perform one-hot encoding on
#I chose 10 as the number of unique values to one hot encode since it is a decent num of cols to make
one_hot_candidates = df.select_dtypes(include=['object','bool']).loc[:, lambda x: x.nunique() < 10]
print(one_hot_candidates.columns.tolist())
```

```python
#run this once
df = pd.get_dummies(df, columns=one_hot_candidates.columns, drop_first=True)
df.head(1)
```

```python
# YOUR CODE HERE
#Fitting and training my logistic regression model
#I am using label_price instead of price since I made a new column for price data without the 1% outlier
X = df[featureList]
y = df['label_price']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1234)
```

```python
sgd = SGDRegressor(random_state=42)
param_grid = {
    'alpha': [0.0001, 0.001, 0.01, 0.1],           # This measures the Regularization strength
    'penalty': ['none', 'l2', 'l1', 'elasticnet'], # Measures the Regularization type
    'max_iter': [1000, 2000],                       # Number of epochs or runs
    'learning_rate': ['constant', 'optimal', 'invscaling', 'adaptive'], 
    'eta0': [0.001, 0.01, 0.1]                      # Initial learning rate
}
```

```python
grid_search = GridSearchCV(
    estimator=sgd,
    param_grid=param_grid,
    cv=5,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
best_sgd = grid_search.best_estimator_
print("Best hyperparameters:", grid_search.best_params_)
```

```python
y_pred = best_sgd.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
print(f"mse:{mse}")
print(f"rmse:{rmse}")
print(f"r^2:{r2}")
```
